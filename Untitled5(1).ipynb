```python
#TASK-1
import pandas as pd
import numpy as np

# Load dataset
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

# Basic data inspection
print("--- Head ---")
print(df.head())

print("\n--- Summary Info ---")
df.info()

print("\n--- Missing Values ---")
print(df.isnull().sum())

# Data Cleaning: Handle missing values
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Grouping analysis
print("\n--- Survival Rate by Gender ---")
print(df.groupby('Sex')['Survived'].mean())
```

```python
# ==========================================
# TASK 2: DATA VISUALIZATION (Matplotlib & Seaborn)
# ==========================================
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 10))

# 1. Bar Chart: Survival Rate by Passenger Class
plt.subplot(2, 2, 1)
sns.barplot(data=df, x='Pclass', y='Survived', palette='Set2')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Class (1 = 1st, 2 = 2nd, 3 = 3rd)')
plt.ylabel('Survival Probability')

# 2. Histogram / KDE: Age Distribution by Survival
plt.subplot(2, 2, 2)
sns.histplot(data=df, x='Age', hue='Survived', kde=True, element='step', palette='coolwarm')
plt.title('Age Distribution by Survival Status')

# 3. Correlation Heatmap
plt.subplot(2, 2, 3)
numeric_df = df.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='Blues', fmt='.2f')
plt.title('Feature Correlation Heatmap')

# 4. Box Plot: Fare vs Passenger Class
plt.subplot(2, 2, 4)
sns.boxplot(data=df, x='Pclass', y='Fare', palette='Pastel1')
plt.yscale('log') # Log scale for better visibility of fare outliers
plt.title('Fare Distribution by Class (Log Scale)')

plt.tight_layout()
plt.show()
```
